```python


with open("hostel_bois.txt", "r", encoding="utf-8") as file:
    lines = file.readlines()

print("Dataset loaded successfully!")
print("Total lines:", len(lines))
```

```python
print("First 10 messages:\n")

for i in range(10):
    print(lines[i].strip())
```

```python
print("=" * 40)
print("WHATSAPP DATASET INFORMATION")
print("=" * 40)

print("Total Lines :", len(lines))
```

```python
# Store parsed messages
messages = []

# Counters
system_count = 0
media_count = 0
deleted_count = 0

# Store unique participants
participants = set()
for line in lines:

    # Remove extra spaces and newline
    line = line.strip()

    # Skip empty lines
    if line == "":
        continue

    # Split timestamp and remaining text
    parts = line.split(" - ", 1)

    if len(parts) != 2:
        system_count += 1
        continue

    timestamp = parts[0]
    rest = parts[1]

    # Check if sender exists
    if ": " not in rest:
        system_count += 1
        continue

    sender, message = rest.split(": ", 1)

    # Count media messages
    if message == "<Media omitted>":
        media_count += 1

    # Count deleted messages
    if message == "This message was deleted":
        deleted_count += 1

    # Store participant
    participants.add(sender)

    # Store message
    messages.append({
        "timestamp": timestamp,
        "sender": sender,
        "message": message
    })
print(messages)
```

```python
print("="*50)
print("CHAT PARSER SUMMARY")
print("="*50)

print("Parsed Messages :", len(messages))
print("Participants    :", len(participants))
print("System Messages :", system_count)
print("Media Messages  :", media_count)
print("Deleted Messages:", deleted_count)
```

```python
# Count messages sent by each participant

message_count = {}

for msg in messages:

    sender = msg["sender"]

    if sender in message_count:
        message_count[sender] += 1
    else:
        message_count[sender] = 1
```

```python
sorted_messages = sorted(message_count.items(), key=lambda x: x[1], reverse=True)

for sender, count in sorted_messages:
    print(sender, ":", count)
print("=" * 60)
print("MESSAGES PER PERSON")
print("=" * 60)

max_messages = sorted_messages[0][1]

for sender, count in sorted_messages:

    percentage = (count / len(messages)) * 100

    bar_length = max(1, int((count / max_messages) * 20))

    bar = "█" * bar_length
    print()

    print(f"{sender:<8} {bar:<20} {count:>4} ({percentage:4.1f}%)")
```

```python
day_count = {}

for msg in messages:

    # timestamp example: "12/04/24, 23:14"
    timestamp = msg["timestamp"]

    # Extract only the date
    date = timestamp.split(", ")[0]

    if date in day_count:
        day_count[date] += 1
    else:
        day_count[date] = 1
```

```python
busiest_day = max(day_count, key=day_count.get)

print("Busiest Day :", busiest_day)
print("Messages    :", day_count[busiest_day])
```

```python
hour_count = {}

for msg in messages:

    timestamp = msg["timestamp"]

    # Extract time
    time = timestamp.split(", ")[1]

    # Extract hour
    hour = time.split(":")[0]

    if hour in hour_count:
        hour_count[hour] += 1
    else:
        hour_count[hour] = 1
busiest_hour = max(hour_count, key=hour_count.get)

print("Busiest Hour :", busiest_hour + ":00 - " + busiest_hour + ":59")
print("Messages     :", hour_count[busiest_hour])
```

```python
participant_list = sorted(list(participants))

print(participant_list)
```

```python
person_index = {}

for i, person in enumerate(participant_list):
    person_index[person] = i

print(person_index)
```

```python
import numpy as np

heatmap = np.zeros((len(participant_list), 24), dtype=int)
for msg in messages:

    sender = msg["sender"]

    timestamp = msg["timestamp"]

    hour = int(timestamp.split(", ")[1].split(":")[0])

    row = person_index[sender]

    heatmap[row][hour] += 1
```

```python
print("=" * 70)
print("ACTIVITY HEATMAP (hour of day, columns 00 to 23)")
print("=" * 70)

# Display only these hours
hours = [0, 3, 6, 9, 12, 15, 18, 21]

print(" " * 10, end="")
for h in hours:
    print(f"{h:02}", end="  ")
print()

for i, person in enumerate(participant_list):

    print(f"{person:<8}", end=" ")

    # Maximum messages of this participant
    max_count = max(heatmap[i])

    for h in hours:

        value = heatmap[i][h]

        if max_count == 0:
            symbol = "."

        else:
            ratio = value / max_count

            if ratio >= 0.75:
                symbol = "█"

            elif ratio >= 0.50:
                symbol = "▒"

            elif ratio >= 0.25:
                symbol = "░"

            else:
                symbol = "."

        print(f"{symbol}  ", end="")

    # Detect Night Owl
    night_messages = (
        heatmap[i][23]
        + heatmap[i][0]
        + heatmap[i][1]
        + heatmap[i][2]
        + heatmap[i][3]
        + heatmap[i][4]
    )

    total_messages = np.sum(heatmap[i])

    if total_messages > 0 and night_messages / total_messages >= 0.75:
        print(" <- NIGHT OWL")
    else:
        print()
```

```python
word_count = {}

for msg in messages:

    text = msg["message"].lower()

    # Skip media and deleted messages
    if text == "<media omitted>" or text == "this message was deleted":
        continue

    words = text.split()

    for word in words:

        # Remove punctuation
        word = word.strip(".,!?;:\"'()[]{}<>")

        if word == "":
            continue

        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1
sorted_words = sorted(word_count.items(), key=lambda x: x[1], reverse=True)
print("=" * 60)
print("THIS GROUP'S FAVOURITE WORDS")
print("=" * 60)

max_count = sorted_words[0][1]

for word, count in sorted_words[:10]:

    bar_length = max(1, int((count / max_count) * 20))

    bar = "█" * bar_length
    print()

    print(f"{word:<8} {bar:<20} {count}")
```

```python
from datetime import datetime, timedelta

print("=" * 60)
print("LONGEST SILENT STREAKS")
print("=" * 60)

# Dataset range
start_date = datetime.strptime("01/04/24", "%d/%m/%y").date()
end_date   = datetime.strptime("30/05/24", "%d/%m/%y").date()

# Active dates for each participant
active_dates = {}

for person in participant_list:
    active_dates[person] = set()

for msg in messages:
    person = msg["sender"]
    d = datetime.strptime(msg["timestamp"], "%d/%m/%y, %H:%M").date()
    active_dates[person].add(d)

# Check silent streak
for person in participant_list:

    longest = 0
    current = 0

    longest_start = None
    longest_end = None

    current_start = None

    day = start_date

    while day <= end_date:

        if day not in active_dates[person]:

            if current == 0:
                current_start = day

            current += 1

            if current > longest:
                longest = current
                longest_start = current_start
                longest_end = day

        else:
            current = 0

        day += timedelta(days=1)

    if longest > 0:
        print(f"{person:<8}: {longest:2} days ({longest_start.strftime('%d %b')} - {longest_end.strftime('%d %b')})")
    else:
        print(f"{person:<8}: 0 days")
```

```python
spam_score = {}

for person in participant_list:

    streaks = []
    streak = 0

    for i in range(len(messages)):

        if messages[i]["sender"] == person:
            streak += 1
        else:
            if streak > 0:
                streaks.append(streak)
            streak = 0

    if streak > 0:
        streaks.append(streak)

    if len(streaks) > 0:
        spam_score[person] = sum(streaks) / len(streaks)
    else:
        spam_score[person] = 0

spammer = max(spam_score, key=spam_score.get)
```

```python
night_score = {}

for i, person in enumerate(participant_list):

    total = np.sum(heatmap[i])

    night = (
        heatmap[i][23]
        + heatmap[i][0]
        + heatmap[i][1]
        + heatmap[i][2]
        + heatmap[i][3]
        + heatmap[i][4]
    )

    if total > 0:
        night_score[person] = (night / total) * 100
    else:
        night_score[person] = 0

night_owl = max(night_score, key=night_score.get)
```

```python
ghost_score = {}

for person in participant_list:
    ghost_score[person] = len(active_dates[person])

ghost = min(ghost_score, key=ghost_score.get)
```

```python
caps_percent = {}

for person in participant_list:

    total = 0
    caps = 0

    for msg in messages:

        if msg["sender"] == person:

            total += 1

            if msg["message"].isupper() and len(msg["message"]) > 2:
                caps += 1

    if total > 0:
        caps_percent[person] = (caps / total) * 100
    else:
        caps_percent[person] = 0

drama = max(caps_percent, key=caps_percent.get)
```

```python
story_score = {}

for person in participant_list:

    words = 0
    msgs = 0

    for msg in messages:

        if msg["sender"] == person:

            words += len(msg["message"].split())
            msgs += 1

    story_score[person] = words / msgs

story = max(story_score, key=story_score.get)
```

```python
keywords = [
    "sleep",
    "eat",
    "take care",
    "study",
    "drink",
    "water",
    "home",
    "care",
    "rest"
]

care_score = {}

for person in participant_list:

    score = 0

    for msg in messages:

        if msg["sender"] == person:

            text = msg["message"].lower()

            for word in keywords:

                if word in text:
                    score += 1

    care_score[person] = score

group_mom = max(care_score, key=care_score.get)
```

```python
print("=" * 60)
print("PERSONALITY ARCHETYPES")
print("=" * 60)

print()

print(f"{spammer:<8} → THE SPAMMER      (avg {spam_score[spammer]:.1f} msgs in a row)")
print(f"{group_mom:<8} → THE GROUP MOM    (care score: {care_score[group_mom]})")
print(f"{night_owl:<8} → THE NIGHT OWL    ({night_score[night_owl]:.1f}% msgs between 23h-04h)")
print(f"{story:<8} → THE STORYTELLER (avg {story_score[story]:.1f} words/msg)")
print(f"{drama:<8} → THE DRAMA QUEEN  ({caps_percent[drama]:.1f}% ALL-CAPS)")
print(f"{ghost:<8} → THE GHOST        (active on {ghost_score[ghost]} days)")

print()
print("=" * 60)
print("Generated by GroupDNA • Built with Python + NumPy")
print("=" * 60)
```
